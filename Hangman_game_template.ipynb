## Description
> ##### The Hangman program randomly selects a secret word from a list of secret words.
> ##### The Game: Here, a random word (a fruit name) is picked up from our collection and the player gets limited chances to guess the fruit name correctly.
> ##### When a letter in that word is guessed correctly, that letter position in the word is made visible in the output.
> ##### In this way, all letters of the word are to be guessed before all the chances are over.
> ##### For convenience, the number of chances equals the word length + 2.
> ##### For example, word to be guessed is mango, then user gets 5 + 2 = 7 chances, as mango is a five letter word.
> ##### Note : Total number of Guesses either right or wrong equals word length + 2.
> ##### Make game repeats until player ends the game.
> ##### Bonus : Create a function that takes 3 parameters `Easy`, `Medium` and `Hard` where `Medium` has more difficult words to be guessed than `Easy` and so on.

```python
import random
```

```python
while True:
    my_list=['apple','banna','mango']
    random_element = list(random.choice(my_list))

    i = 0
    list_loc=['-'] * len(random_element)
    while i < len(random_element) + 2:
        # if list_loc == random_element:
        #     print("Congratulations, you won!")
        #     break
        # else:
        print(list_loc)
        guess = input('guess a char!')
        for j in range(len(random_element)):
            if random_element[j] == guess:
                list_loc[j] = guess

        i = i + 1
    print(list_loc)
    print(''.join(list_loc))
    if list_loc == random_element:
        print("Congratulations, you won!")
    else:
        pass
    play_agian=input("Do you want to play agian [yes/no]:").lower()
    if play_agian in ['yes','y']:
        continue
    else:
        print('Thanks for playing')
        break


```

```python
def play(easy_level, medium_level, hard_level):
    while True:
        level = input('Enter the level you wanted [easy/medium/hard]').lower()
        
        dic = {
            'easy': easy_level,
            'medium': medium_level,
            'hard': hard_level
        }

        if level not in dic:
            print('Enter a vaild level!')
        else:   
            random_element = list(random.choice(dic[level]).replace(' ',''))
            i = 0
            list_loc=['-'] * len(random_element)
            while i < len(random_element) + 2:
                print(list_loc)
                guess = input('guess a char!')
                for j in range(len(random_element)):
                    if random_element[j] == guess:
                        list_loc[j] = guess

                i = i + 1
            print(list_loc)
            # print(''.join(list_loc))
            if list_loc == random_element:
                print("Congratulations, you won!")
            else:
                pass
            play_agian=input("Do you want to play agian [yes/no]:").lower()
            if play_agian in ['yes','y']:
                continue
            else:
                print('Thanks for playing')
                break


```

```python
level_1 = ['apple', 'grape', 'mango', 'peach', 'banna']
level_2 = ['blueberry', 'apricot', 'papaya', 'coconut', 'orange']
level_3 = ['pomegranate', 'dragon fruit', 'passion fruit', 'mangosteen']


play(level_1, level_2, level_3)
```
